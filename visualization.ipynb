```python
import numpy as np

df.columns = [str(c).strip() for c in df.columns]

for c in df.columns:
    if df[c].dtype == object:
        df[c] = df[c].astype(str).str.strip()

# ép các cột có dạng số về numeric
for c in df.columns:
    if df[c].dtype == object:
        s = df[c].str.replace(",", "", regex=False)
        converted = pd.to_numeric(s, errors="coerce")
        non_empty = s.replace("", np.nan).notna().sum()
        if non_empty > 0 and converted.notna().sum() >= max(1, int(0.7 * non_empty)):
            df[c] = converted

dup_rows = df[df.duplicated(subset=["ThA", "ThW"], keep="first")].sort_values(["ThA", "ThW"])

print("===== Các dòng bị trùng theo (ThA, ThW) =====")
print(dup_rows if len(dup_rows) > 0 else "Không có dòng trùng")

print("\nSố dòng trước khi xoá trùng:", len(df))

# giữ dòng đầu tiên, xoá các dòng trùng còn lại
df = df.drop_duplicates(subset=["ThA", "ThW"], keep="first").reset_index(drop=True)

print("Số dòng sau khi xoá trùng:", len(df))

INITIAL_NUM_EDGES = 163_788
INITIAL_TENSOR_NUMEL = 327_576   # = 2 * INITIAL_NUM_EDGES

# NumelA/NumelW trong log đang ở đơn vị nghìn
df["adj_numel"] = pd.to_numeric(df["NumelA"], errors="coerce") * 1000
df["weight_numel"] = pd.to_numeric(df["NumelW"], errors="coerce") * 1000

# số cạnh còn lại sau prune adjacency
df["edge_count"] = df["adj_numel"] / 2

# sparsity
df["adj_sparsity"] = 1 - df["adj_numel"] / INITIAL_TENSOR_NUMEL

INITIAL_NUMEL_W = df["weight_numel"].max()
df["weight_sparsity"] = 1 - df["weight_numel"] / INITIAL_NUMEL_W

# FLOPs = 2 * MAC
df["FLOPs_train"] = 2 * pd.to_numeric(df["Ctrain"], errors="coerce")
df["FLOPs_test"]  = 2 * pd.to_numeric(df["CTest"], errors="coerce")

# Sweep theo thr_a khi thr_w cố định = 0
df_adj_sweep = (
    df[df["ThW"] == 0]
    .sort_values("ThA")
    .reset_index(drop=True)
)

# Sweep theo thr_w khi thr_a cố định = 0
df_weight_sweep = (
    df[df["ThA"] == 0]
    .sort_values("ThW")
    .reset_index(drop=True)
)
```

```python
import matplotlib.pyplot as plt

# 1. Accuracy theo ThA
plt.figure(figsize=(8, 5))
plt.plot(df_adj_sweep["ThA"], df_adj_sweep["Acc"], marker="o", linewidth=2)
plt.xlabel("ThA")
plt.ylabel("Accuracy")
plt.title("Accuracy vs ThA (ThW = 0)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# 2. Accuracy theo ThW
plt.figure(figsize=(8, 5))
plt.plot(df_weight_sweep["ThW"], df_weight_sweep["Acc"], marker="o", linewidth=2)
plt.xlabel("ThW")
plt.ylabel("Accuracy")
plt.title("Accuracy vs ThW (ThA = 0)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# 3. Accuracy theo adjacency sparsity
plt.figure(figsize=(8, 5))
plt.plot(df_adj_sweep["adj_sparsity"], df_adj_sweep["Acc"], marker="o", linewidth=2)
plt.xlabel("Adjacency Sparsity")
plt.ylabel("Accuracy")
plt.title("Accuracy vs Adjacency Sparsity")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# 4. Accuracy theo weight sparsity
plt.figure(figsize=(8, 5))
plt.plot(df_weight_sweep["weight_sparsity"], df_weight_sweep["Acc"], marker="o", linewidth=2)
plt.xlabel("Weight Sparsity")
plt.ylabel("Accuracy")
plt.title("Accuracy vs Weight Sparsity")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# 5. Accuracy theo FLOPs test cho adjacency sweep
plt.figure(figsize=(8, 5))
plt.plot(df_adj_sweep["FLOPs_test"], df_adj_sweep["Acc"], marker="o", linewidth=2)
plt.xlabel("Test FLOPs")
plt.ylabel("Accuracy")
plt.title("Accuracy vs Test FLOPs (Adj Sweep)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# 6. Accuracy theo FLOPs test cho weight sweep
plt.figure(figsize=(8, 5))
plt.plot(df_weight_sweep["FLOPs_test"], df_weight_sweep["Acc"], marker="o", linewidth=2)
plt.xlabel("Test FLOPs")
plt.ylabel("Accuracy")
plt.title("Accuracy vs Test FLOPs (Weight Sweep)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
